```python
import digitalhub as dh

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")


# ---- carico la lista delle 5000 serie come artifact, nel progetto "datasets" ----
project_data = dh.get_project("datasets")
project_data.log_artifact(
    name="nomi_serie_5000_water100_bilanciato",
    source="../../csv/nomi_serie_5000_water100_bilanciato.txt",   # percorso locale del file
    kind="artifact",
)
print("OK -> lista serie caricata")


# ---- parametri ----
job_name = "v1"

parameters = {
    "series_list_artifact": "nomi_serie_5000_water100_bilanciato",
    "anomalies_artifact": "Floods_Anomalies",
    "splits": ["train", "val", "test"],
    "output_name": "Floods_Anomalies_crop_norm",
    "max_part_gb": 15.0,
    "max_series": 0,            # 0 = tutte; N > 0 = solo le prime N (ordine alfabetico) per una prova veloce
    "rgb_gain": 2.5,          # guadagno fisso della normalizzazione OPT_RGB
    "rgb_gamma": 2.2,         # gamma della normalizzazione OPT_RGB (1.0 = lineare, piu' scuro)
    "include_mask": True,  
    "invalid_as_zero": True,
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",
        "spec": {"size": "500Gi"}
    }
]

# ---- build ----
crop_norm_anom_func = project.new_function(
    name=f"crop_norm_anom_{job_name}",
    kind="python",
    python_version="PYTHON3_10",
    code_src="../../src/",
    handler="utils:crop_norm_artifact",
    base_image="pytorch/pytorch:2.1.2-cuda11.8-cudnn8-runtime",
    requirements=["pandas==2.3.3", "numpy==1.26.4", "rasterio==1.4.4", "tqdm==4.70.0", "tifffile==2024.8.30", "torch==2.1.2", "matplotlib==3.10.9", "digitalhub==0.15.11", "digitalhub-runtime-python==0.15.2", "zarr==2.18.0", "numcodecs==0.13.1", "xarray==2025.6.1"]

)

build = crop_norm_anom_func.run("build", wait=True)
print(f"BUILD: {build.status.state}")

# ---- esecuzione: nessun torch nell'handler, niente GPU necessaria ----
run_crop_norm_anom = crop_norm_anom_func.run(
    action="job",
    parameters=parameters,
    volumes=volumes,
    profile="cpu",
    local_execution=False,
    wait=True
)

print(f"Run avviato: {run_crop_norm_anom.id}")
```
