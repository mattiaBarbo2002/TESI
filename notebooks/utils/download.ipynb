```python
import digitalhub as dh

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")

job_name = "dl_impactmesh"

parameters = {
    "repo_id": "ibm-esa-geospatial/ImpactMesh-Flood",
    "split": None,
    "output_name": "Floods_Anomalies_v2",
    "max_workers": 8,
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",
        "spec": {"size": "600Gi"}
    }
]

download_func = project.new_function(
    name=f"download_impactmesh_{job_name}",
    kind="python",
    python_version="PYTHON3_10",
    code_src="../src/",
    handler="download_impactmesh_split",
    base_image="pytorch/pytorch:2.1.2-cuda11.8-cudnn8-runtime",
    requirements=["pandas==2.3.3", "numpy==1.26.4", "rasterio==1.4.4", "tqdm==4.70.0", "tifffile==2024.8.30", "torch==2.1.2", "matplotlib==3.10.9", "digitalhub==0.15.11", "digitalhub-runtime-python==0.15.2", "zarr==2.18.0", "numcodecs==0.13.1", "xarray==2025.6.1", "huggingface_hub==1.33.0"]
)

build = download_func.run("build", wait=True)
print(f"BUILD: {build.status.state}")

run_download = download_func.run(
    action="job",
    parameters=parameters,
    volumes=volumes,
    profile="cpu",
    local_execution=False,
    wait=True
)

print(f"Run avviato: {run_download.id}")
```
