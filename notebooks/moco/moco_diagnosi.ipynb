**MOCO EVAL**

```python
import digitalhub as dh

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

**SETUP PARAMETERS**

```python
                                                                                                           

parameters = {
    "moco_version": "dense_v4",
    "dataset": "Test",
    "hidden_channels_dim": 128,
    "pool_grid": 4,
    "moco_dim": 128,
    "moco_k": 4096,
    "grid_dense": 8,
    "T_dense": 0.2,
    "n_neg": 4,
    "batch_size": 16,
    "n_batch": 20,
    "patch_size": 256,
    "n_images1": 4, "n_channels1": 2,
    "n_images2": 4, "n_channels2": 10,
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",      
        "spec": {"size": "100Gi"}   
    }
]
```

**BUILD ENVIRONMENT**

```python
moco_func = project.new_function(
    name=f'diagnosi',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../../src/",
    handler="moco:diagnosi_dense",
    image="harbor.atlas.fbk.eu/rsde/rsde-platform-floods-floodsimagev3:9b7e2",
)
```

**TRAINING**

```python
resources = {"cpu": "4", "mem": "10Gi"}

run_moco_2D = moco_func.run(
    action="job", 
    parameters=parameters, 
    volumes=volumes, 
    resources=resources,
    profile="1xv100",
    local_execution= False,                       
    wait=True
)

print(f"Run moco avviato: {run_moco_2D.id}")
```
