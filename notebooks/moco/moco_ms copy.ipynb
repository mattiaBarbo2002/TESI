**MOCO 2D**

```python
import digitalhub as dh
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

**SETUP PARAMETERS**

```python
# lr = 0.001, lr_min = 0.00005
                                        

parameters = {
    "job_name": "net_v5_ms_ep10pre",
    "init_from": "2D_weights_net_v5",
    "init_version": "9ccf57b762264d468e2a6fd400e21fb9",


    "resume": False,
    "hidden_channels_dim": 128,
    "pool_grid": 4,
    "moco_k": 4096,
    "lr": 0.002,
    "cosine": True, "lr_min": 0.0001,
    "epochs": 30,  
    "center_m": 1.0,              
}


volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",      
        "spec": {"size": "400Gi"}   
    }
]
```

**BUILD ENVIRONMENT**

```python
moco_func = project.new_function(
    name=f'net_v5_ms_ep10pre',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../../src/",
    handler="moco:train_moco_ms",
    image="harbor.atlas.fbk.eu/rsde/rsde-platform-floods-floodsimagev3:9b7e2",
)
```

**TRAINING**

```python
resources = {"cpu": "4", "mem": "19Gi"}

run_moco_2D = moco_func.run(
    action="job", 
    parameters=parameters, 
    volumes=volumes, 
    resources=resources,
    profile="1xv100-shared",
    local_execution= False,                       
    wait=True
)

print(f"Run moco avviato: {run_moco_2D.id}")
```

**PLOTS**

```python
path = project.get_artifact(f"moco_bot_metrics_{job_name}").download(overwrite=True)
df = pd.read_csv(path)

plt.figure(figsize=(8, 5))
plt.plot(df['epoch'], df['train_loss'], color='green', label='Train Loss')
plt.title(f'MoCo 2D - {job_name}')
plt.xlabel('Epochs')
plt.ylabel('Loss InfoNCE')
plt.grid(True)
plt.legend()
plt.show()
```
