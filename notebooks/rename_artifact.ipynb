**RENAME ARTIFACT**

```python
import digitalhub as dh

NOME_PROGETTO = "floods"
project = dh.get_project(NOME_PROGETTO)
print(f"Progetto: {project.name}")
```

```python
job_name = "moco_2D_net"

parametri = {
    "old_metrics": "moco-metrics_moco_2D_convLstm_net_v4_Standard_200",
    "old_weights": "moco-weights_moco_2D_convLstm_net_v4_Standard_200",
    "new_metrics": "moco_2D_metrics_net_v1",
    "new_weights": "moco_2D_weights_net_v1",                
}

volumi = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",      
        "spec": {"size": "10Gi"}   
    }
]
```

```python
rename_artifact_func = project.new_function(
    name= f'rename_artifact_{job_name}',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../src/", 
    handler="rename_artifact", 
    base_image="pytorch/pytorch:2.1.2-cuda11.8-cudnn8-runtime",
    requirements=["pandas==2.3.3", "numpy==1.26.4", "rasterio==1.4.4", "tqdm==4.70.0", "tifffile==2024.8.30", "torch==2.1.2", "matplotlib==3.10.9", "digitalhub==0.15.11", "digitalhub-runtime-python==0.15.2"]
)

build = rename_artifact_func.run("build", wait=True)
print(f"BUILD: {build.status.state}")

run_rename_artifact = rename_artifact_func.run(
    action="job", 
    parameters=parametri, 
    volumes=volumi, 
    profile="1xv100-shared",                                       
    local_execution= False,
    wait=True                               
)

print(f"Run train_encoders avviato: {run_rename_artifact.id}")
```
