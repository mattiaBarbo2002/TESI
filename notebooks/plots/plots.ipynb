**PLOTS**

```python
import digitalhub as dh
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

```python
job_names = ["autoencoder_2D_SAR_metrics_convLstm_v1", "autoencoder_attn_SAR_metrics_v1", "autoencoder_attn_SAR_metrics_v2"]
dimensions = ["2d", "attn_v1", "attn_v2"]
colors = ["orange", "green", "blue"]

plt.figure(figsize=(8, 5))

dfs = {}
for job_name, dim in zip(job_names, dimensions):
    path = project.get_artifact(job_name).download(overwrite=True)
    dfs[dim] = pd.read_csv(path)

max_epoch = min(df['epoch'].max() for df in dfs.values())

plt.figure(figsize=(8, 5))
for dim, color in zip(dimensions, colors):
    df = dfs[dim]
    df = df[df['epoch'] <= max_epoch]
    plt.plot(df['epoch'], df['train_loss'], color=color, label=dim)

plt.title('SAR')
plt.xlabel('Epochs')
plt.ylabel('Loss MSE')
plt.grid(True)
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2)
plt.tight_layout()
plt.show()
```
