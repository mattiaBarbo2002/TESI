**ANOMALY DETECTION OC-SVM**

```python
import digitalhub as dh
import zipfile
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

**SETUP PARAMETERS**

```python

parameters = {
    "job_name": "linReg_v5",
    "versions": [
        {"name": "v5", "results_artifact": "anomaly_results_ocsvm_net_v5_0per_postnorm",
         "moco_attn": False, "moco_version": "net_v5", "pool_grid": 4, "attn": False, "ltae": False},
    ],
    "hidden_channels_dim": 128,
    "include_normals": True,
    "save_features": True,
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",
        "spec": {"size": "300Gi"}
    }
]
```

**BUILD ENVIRONMENT**

```python
ocsvm_func = project.new_function(
    name=f'lin_reg_v5',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../../src/",
    handler="anomaly:linear_regression",
    image="harbor.atlas.fbk.eu/rsde/rsde-platform-floods-floodsimagev3:9b7e2",
)
```

**TRAINING**

```python
resources = {"cpu": "4", "mem": "50Gi"}

run_ad = ocsvm_func.run(
    action="job",
    parameters=parameters,
    volumes=volumes,
    resources = resources,
    profile="1xv100-shared",
    local_execution=False,
    wait=True
)

print(f"Run anomaly detection ocsvm: {run_ad.id}")
```
