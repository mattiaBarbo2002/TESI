**ANOMALIES EXTRACTION**

```python
import digitalhub as dh

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

**SETUP PARAMETERS**

```python
parameters = {
    "split": "test",          
    "n_series_sample": 100,   #4974  
    "seed": 0,
    "max_part_gb": 15.0,
    # "n_show": 30,
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",      
        "spec": {"size": "500Gi"}   
    }
]
```

**BUILD ENVIRONMENT**

```python
extract_anomalies_func = project.new_function(
    name= f'extract_anomalies',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../src/", 
    handler="build_precomputed_cache_anomalies", 
    base_image="pytorch/pytorch:2.1.2-cuda11.8-cudnn8-runtime",
    requirements=["pandas==2.3.3", "numpy==1.26.4", "rasterio==1.4.4", "tqdm==4.70.0", "tifffile==2024.8.30", "torch==2.1.2", "matplotlib==3.10.9", "digitalhub==0.15.11", "digitalhub-runtime-python==0.15.2", "zarr==2.18.0", "numcodecs==0.13.1", "xarray==2025.6.1"]
)

build = extract_anomalies_func.run("build", wait=True)
print(f"BUILD: {build.status.state}")
```

**TRAINING**

```python
run_extract_anomalies = extract_anomalies_func.run(
    action="job", 
    parameters=parameters, 
    volumes=volumes, 
    profile="cpu",
    local_execution= False,                       
    wait=True
)

print(f"Run moco avviato: {run_extract_anomalies.id}")
```
