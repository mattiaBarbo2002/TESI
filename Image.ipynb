```python
import digitalhub as dh

PROJECT_NAME = "floods"
project = dh.get_project(PROJECT_NAME)
print(f"Project: {project.name}")
```

**BUILD IMMAGINE**

```python
build_image = project.new_function(
    name=f'FloodsImage_v3',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../src/",
    handler="utils:FloodsImage",
    base_image="pytorch/pytorch:2.1.2-cuda11.8-cudnn8-runtime",
    requirements=["pandas==2.3.3", "numpy==1.26.4", "rasterio==1.4.4", "tqdm==4.70.0", "tifffile==2024.8.30", "torch==2.1.2", "matplotlib==3.10.9", "digitalhub==0.15.11", "digitalhub-runtime-python==0.15.2", "zarr==2.18.0", "numcodecs==0.13.1", "xarray==2025.6.1", "scikit-learn==1.5.2"]
)

build = build_image.run("build", wait=True)
```

**USO**

```python
parameters = {
    "name": "prova",
    "version": "v3",
}

volumes = [
    {
        "volume_type": "ephemeral",
        "name": "volume-spazio-dati",
        "mount_path": "/data",      
        "spec": {"size": "10Gi"}   
    }
]

resources = {
    "cpu": "4", 
    "mem": "20Gi",
}
```

```python
func_name = project.new_function(
    name=f'name',
    kind="python",
    python_version="PYTHON3_10",
    code_src="../src/",
    handler="utils:FloodsImage",
    image="harbor.atlas.fbk.eu/rsde/rsde-platform-floods-floodsimagev3:9b7e2"
)


print(f"BUILD: {build.status.state}")
```

```python
run_func = func_name.run(
    action="job",
    parameters=parameters,
    volumes=volumes,
    resources = resources,
    profile="1xv100-shared",
    local_execution=False,
    wait=True
)

print(f"Run avviato: {run_func.id}")
```
